```python
import pandas as pd

# 读取原始CSV文件
data = pd.read_csv(r'/Users/dc/Z研究生/eedsProject/device_10001_sorted.csv')
data["device_id"].unique()
```

```python



# 删除最后三列
# 方法1: 使用iloc选择所有列，除了最后三列
data_modified = data.iloc[:, :-3]

# 或者方法2: 直接指定要保留的列（除了最后三列）
# columns_to_keep = data.columns[:-3]
# data_modified = data[columns_to_keep]

# 保存修改后的数据到新的CSV文件
data_modified.to_csv(r'/Users/dc/Z研究生/eedsProject/device_10001_sorted_modified.csv', index=False)

print("已删除最后三列并保存到新文件")
print("原始列数:", len(data.columns))
print("修改后列数:", len(data_modified.columns))
print("删除的列:", list(data.columns[-3:]))
```

```python
data['area_id_seq'].unique()
gx_min, gx_max = data['grid_x'].min(), data['grid_x'].max()
gy_min, gy_max = data['grid_y'].min(), data['grid_y'].max()
print(f'grid_x: {gx_min} ~ {gx_max}')
print(f'grid_y: {gy_min} ~ {gy_max}')

# 如果连续，则 最大值 - 最小值 + 1 == 唯一值数量
gx_min, gx_max = data['grid_x'].min(), data['grid_x'].max()
unique_count = data['grid_x'].nunique()

is_continuous = (gx_max - gx_min + 1) == unique_count
print(f'grid_x 范围：{gx_min} ~ {gx_max}')
print(f'唯一值数量：{unique_count}')
print(f'是否连续：{is_continuous}')
```
